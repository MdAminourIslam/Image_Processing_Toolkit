```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

print("OpenCV version:", cv2.__version__)
```

```python
image = cv2.imread("cameraman.tif")

print("Image shape:", image.shape)
```

```python
# Convert BGR image to RGB for correct display
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8, 6))
plt.imshow(image_rgb)
plt.axis("off")
plt.show()
```

```python
# Resize the image
resized = cv2.resize(image, (640, 360))

print("Original:", image.shape)
print("Resized:", resized.shape)
```

```python
# Crop a region from the image
cropped = image[100:500, 200:700]

cropped_rgb = cv2.cvtColor(cropped, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8, 5))
plt.imshow(cropped_rgb)
plt.axis("off")
plt.show()
```

```python
# Convert the image to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

print("Grayscale shape:", gray.shape)
```

```python
plt.figure(figsize=(8, 6))
plt.imshow(gray, cmap="gray")
plt.axis("off")
plt.show()
```

```python
# Apply Gaussian blur to reduce noise
blurred = cv2.GaussianBlur(gray, (9, 9), 0)

plt.figure(figsize=(8, 6))
plt.imshow(blurred, cmap="gray")
plt.axis("off")
plt.show()
```

```python
# Convert grayscale image into a binary image
_, threshold = cv2.threshold(
    gray,
    127,
    255,
    cv2.THRESH_BINARY
)

plt.figure(figsize=(8, 6))
plt.imshow(threshold, cmap="gray")
plt.axis("off")
plt.show()
```

```python
# Detect important edges in the image
edges = cv2.Canny(gray, 100, 200)

plt.figure(figsize=(8, 6))
plt.imshow(edges, cmap="gray")
plt.axis("off")
plt.show()
```

```python
# Reduce noise before edge detection
blurred = cv2.GaussianBlur(gray, (5, 5), 0)

# Detect edges
edges = cv2.Canny(blurred, 100, 200)

plt.figure(figsize=(8, 6))
plt.imshow(edges, cmap="gray")
plt.axis("off")
plt.show()
```

```python
# Display all major image-processing results
plt.figure(figsize=(15, 10))

plt.subplot(2, 3, 1)
plt.imshow(image_rgb)
plt.title("Original")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(cv2.cvtColor(resized, cv2.COLOR_BGR2RGB))
plt.title("Resized")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(cropped_rgb)
plt.title("Cropped")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(threshold, cmap="gray")
plt.title("Threshold")
plt.axis("off")

plt.subplot(2, 3, 6)
plt.imshow(edges, cmap="gray")
plt.title("Canny Edges")
plt.axis("off")

plt.show()
```

```python
# Compare different Gaussian blur kernel sizes
blur_3 = cv2.GaussianBlur(gray, (3, 3), 0)
blur_7 = cv2.GaussianBlur(gray, (7, 7), 0)
blur_11 = cv2.GaussianBlur(gray, (11, 11), 0)

# Compare different Canny thresholds
edge_1 = cv2.Canny(gray, 50, 150)
edge_2 = cv2.Canny(gray, 100, 200)
edge_3 = cv2.Canny(gray, 150, 250)

# Display the results
fig, axes = plt.subplots(2, 3, figsize=(15, 9))

images = [
    (blur_3, "Blur 3x3"),
    (blur_7, "Blur 7x7"),
    (blur_11, "Blur 11x11"),
    (edge_1, "Canny 50-150"),
    (edge_2, "Canny 100-200"),
    (edge_3, "Canny 150-250")
]

for ax, (img, title) in zip(axes.flat, images):
    ax.imshow(img, cmap="gray")
    ax.set_title(title)
    ax.axis("off")

plt.tight_layout()
plt.show()
```
